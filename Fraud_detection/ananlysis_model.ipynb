--- FEATURE ENGINEERING & TRAINING A ML MODEL ---

```python
# Data splitting & preprocessing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Models
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

# Evaluation
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

# Saving models
import joblib
```

```python
df.head()
```

```python
df_model = df.drop(["nameOrig", "nameDest", "isFlaggedFraud"], axis=1)
```

```python
df_model.head()
```

```python
categorial = ["type"]
numeric = ["amount", "oldbalanceOrg", "newbalanceOrig", "oldbalanceDest", "newbalanceDest"]
```

```python
y = df_model["isFraud"]
X = df_model.drop("isFraud", axis=1)
```

```python
X_train_full, X_test_full, y_train_full, y_test_full = train_test_split(
    X, y, test_size=0.3, stratify=y, random_state=42
)
```

```python
preprocessor = ColumnTransformer(
    transformers= [
        ("num", StandardScaler(), numeric),
        ("cat", OneHotEncoder(drop="first"), categorial)
    ],
    remainder="drop"
)
```

```python
pipeline = Pipeline([
    ("prep", preprocessor),
    ("clif", LogisticRegression(class_weight="balanced", max_iter=1000))
])
```

```python
pipeline.fit(X_train_full, y_train_full)
```

```python
y_pred = pipeline.predict(X_test_full)
```

```python
print(classification_report(y_test_full, y_pred))
```

```python
confusion_matrix(y_test_full, y_pred)
```

--- Train Random Forest and XGBoost models on 5% data sample ---

```python
# Take a 5% sample for ensemble models
df_sample = df_model.sample(frac=0.05, random_state=42)
X_sample = df_sample.drop("isFraud", axis=1)
y_sample = df_sample["isFraud"]
```

```python
X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(
    X_sample, y_sample, test_size=0.3, stratify=y_sample, random_state=42
)
```

```python
# Define ensemble models
models_sample = {
    "Random Forest": RandomForestClassifier(
        class_weight="balanced", n_estimators=100, max_depth=10, n_jobs=-1, random_state=42
    ),
    "XGBoost": XGBClassifier(
        n_estimators=100,
        max_depth=5,
        scale_pos_weight=(y_train_s.value_counts()[0] / y_train_s.value_counts()[1]),
        use_label_encoder=False,
        eval_metric="logloss",
        n_jobs=-1,
        random_state=42
    )
}
```

```python
best_model_name = None
best_roc_auc = 0
best_pipeline = None
```

```python
trained_pipelines = {}

for name, model in models_sample.items():
    pipeline = Pipeline([
        ("prep", preprocessor),
        ("clf", model)
    ])
    
    pipeline.fit(X_train_s, y_train_s)
    trained_pipelines[name] = pipeline  # ✅ save trained pipeline
    
    y_pred = pipeline.predict(X_test_s)
    y_proba = pipeline.predict_proba(X_test_s)[:, 1]
    
    print(f"=== {name} (5% sample) ===")
    print(classification_report(y_test_s, y_pred))
    print("Confusion Matrix:\n", confusion_matrix(y_test_s, y_pred))
    roc_auc = roc_auc_score(y_test_s, y_proba)
    print("ROC-AUC:", roc_auc)
    print("-"*50)
```

```python
# Collect metrics for all models
comparison = []

# Logistic Regression (full dataset)
y_pred_lr = pipeline.predict(X_test_full)
y_proba_lr = pipeline.predict_proba(X_test_full)[:, 1]
comparison.append({
    "Model": "Logistic Regression (Full)",
    "Precision": classification_report(y_test_full, y_pred_lr, output_dict=True)["1"]["precision"],
    "Recall": classification_report(y_test_full, y_pred_lr, output_dict=True)["1"]["recall"],
    "F1-score": classification_report(y_test_full, y_pred_lr, output_dict=True)["1"]["f1-score"],
    "ROC-AUC": roc_auc_score(y_test_full, y_proba_lr)
})

# Random Forest and XGBoost (5% sample)
for name, pipeline in trained_pipelines.items():
    y_pred = pipeline.predict(X_test_s)
    y_proba = pipeline.predict_proba(X_test_s)[:, 1]
    
    comparison.append({
        "Model": f"{name} (5% Sample)",
        "Precision": classification_report(y_test_s, y_pred, output_dict=True)["1"]["precision"],
        "Recall": classification_report(y_test_s, y_pred, output_dict=True)["1"]["recall"],
        "F1-score": classification_report(y_test_s, y_pred, output_dict=True)["1"]["f1-score"],
        "ROC-AUC": roc_auc_score(y_test_s, y_proba)
    })


# Convert to DataFrame
comparison_df = pd.DataFrame(comparison)
comparison_df
```

```python
metrics = ["Precision", "Recall", "F1-score", "ROC-AUC"]

fig, axes = plt.subplots(1, 4, figsize=(18,5), sharey=True)

for i, metric in enumerate(metrics):
    axes[i].bar(comparison_df.index, comparison_df[metric], color=["#1f77b4","#ff7f0e","#2ca02c"])
    axes[i].set_title(metric, fontsize=12, weight="bold")
    axes[i].set_ylim(0, 1.05)
    axes[i].tick_params(axis="x", rotation=45)

fig.suptitle("Model Performance Comparison", fontsize=16, weight="bold")
plt.tight_layout()
plt.show()
```
